## Max Absolute Scaling

Max Absolute Scaling is a feature scaling technique that divides each value by the maximum absolute value of its feature. It scales the data to a range between **−1 and 1** while preserving the signs of the original values.

**Formula:**

$$
X' = \frac{X}{|X|_{\max}}
$$

Where:
- $X$ = Original value
- $|X|_{\max}$ = Maximum absolute value of the feature

**Key observations:**
- The maximum absolute value of each feature becomes 1 (unless the feature contains only zeros).
- Positive and negative values retain their signs.
- It does not center the data around zero by subtracting the mean.
- It preserves zero values and is useful for sparse datasets.
