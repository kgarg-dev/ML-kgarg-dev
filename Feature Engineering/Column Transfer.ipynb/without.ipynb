## Encoding , filling missing values , concatenation  ##

**Encoding used :-**
1. Label encoder for labels(y)
2. Ordinal encoder for ordinal data column
3. One hot encoding for Nominal data column

**Handling missing values**
1. could have been done by fillna()
2. Performed by scikit by importing Simple imputer from scikit.impute

**concatenation**
1. method 1:- np.concatenate
2. method 2:- np.hstack

## TRAIN_TEST_SPLIT ##

```python
from sklearn.model_selection import train_test_split

X_train , X_test , y_train , y_test = train_test_split(df.drop(columns=['has_covid']) , df['has_covid'] , test_size=0.2 , random_state=47)
```

```python
X_train.sample(3)
```

## Encoding ##

**we will do one hot encoding on nomial columns** 

```python
from sklearn.preprocessing import OneHotEncoder
oe=OneHotEncoder(sparse_output=False , drop='first' , dtype=np.int32)

X_train_oe=oe.fit_transform(X_train[['gender' , 'city']])
X_test_oe=oe.transform(X_test[['gender' , 'city']])

X_train_oe.shape
# 3 columns of city and 1 of gender :- as 1 from each dropped so 4 columns in total
```

**filling out missing values in fever**

```python
from sklearn.impute import SimpleImputer

#fills mean in Null values

imputer=SimpleImputer()
X_train_fever=imputer.fit_transform(X_train[['fever']])
X_test_fever=imputer.transform(X_test[['fever']])

X_train_fever
```

```python
X_train_fever.shape
```

**Ordinal encoding**

```python
from sklearn.preprocessing import OrdinalEncoder
ore=OrdinalEncoder(categories=[['Mild','Strong']])

X_train_cough = ore.fit_transform(X_train[['cough']])
X_test_cough = ore.transform(X_test[['cough']])

X_train_cough
```

**Label encoding**

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()

y_train_label=le.fit_transform(y_train)
y_test_label=le.transform(y_test)

y_train_label
```

```python
#conversion to dataframe
y_train_label=pd.DataFrame(y_train_label , columns=['has_covid'])
y_test_label=pd.DataFrame(y_test_label , columns=['has_covid'])

y_train_label.head()
```

**Extracting X_train_age**

```python
X_train_age = X_train[['age']].values   
X_test_age = X_test[['age']].values

X_train_age.shape
```

## Concatenation ##

```python
#method 1
X_train_transformed=np.concatenate((X_train_age,X_train_fever,X_train_oe,X_train_cough),axis=1)
X_test_transformed=np.concatenate((X_test_age,X_test_fever,X_test_oe,X_test_cough),axis=1)

X_train_transformed
```
